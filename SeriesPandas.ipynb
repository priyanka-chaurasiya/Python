```python
import pandas as pd

x=[3,4,5,6,7,8]

var = pd.Series(x)
print(var)
print(type(var))
```

```python
import pandas as pd

x=[3,4,5,6,7,8]

var = pd.Series(x)
print(var)
print(type(var))
print(var[2])
```

```python
import pandas as pd

x=[3,4,5,6]

var = pd.Series(x,index=['a','s','d','f'],dtype="float")
print(var)
print(type(var))
print(var[2])
```

```python
import pandas as pd

x=[3,4,5,6]

var = pd.Series(x,index=['a','s','d','f'],dtype="float",name="python")
print(var)
print(type(var))
print(var[2])
```

```python
dic = {"name": ['python','c','c++','java'],"por":[12,13,14,15],"rank":[1,4,3,2]}

var1 = pd.Series(dic)
print(var1)
```

```python
s = pd.Series(12,index=[1,2,3,4,5,6])
print(s)
print(type(s))
```

```python
s1 = pd.Series(12,index=[1,2,3,4,5,6,7])
s2 = pd.Series(12,index=[1,2,3,4])
print(s1+s2)
```
