```python
# Groupby
```

```python
import pandas as pd

var = pd.DataFrame({"name":["a","b","c","d","a","b","a","b","a","c","c","c"],
                    "S_1":[12,13,14,12,13,15,43,25,16,23,44,65],
                    "S_2":[12,13,14,12,13,15,43,25,16,23,43,55]})

var
```

```python
var_new = var.groupby("name")

var_new
```

```python
for x,y in var_new:
  print(x)
  print(y)
  print()
```

```python
var_new.get_group("a")
```

```python
var_new.get_group("d")
```

```python
var_new.min()
```

```python
var_new.max()
```
