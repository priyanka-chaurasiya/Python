Important CSV + Pandas Functions Cheat Sheet



| Function               | Easy Definition         |
| ---------------------- | ----------------------- |
| `pd.read_csv()`        | CSV read karna          |
| `df.to_csv()`          | CSV save karna          |
| `df.head()`            | Starting rows           |
| `df.tail()`            | Last rows               |
| `df.info()`            | Data information        |
| `df.describe()`        | Statistics              |
| `df.shape`             | Rows + columns          |
| `df.columns`           | Column names            |
| `df.dtypes`            | Data types              |
| `df.loc[]`             | Label se data select    |
| `df.iloc[]`            | Position se data select |
| `df.isnull()`          | Missing values check    |
| `df.dropna()`          | Missing rows delete     |
| `df.fillna()`          | Missing values fill     |
| `df.duplicated()`      | Duplicate check         |
| `df.drop_duplicates()` | Duplicate delete        |
| `df.rename()`          | Name change             |
| `df.drop()`            | Row/column delete       |
| `df.insert()`          | Column insert           |
| `df.sort_values()`     | Data sorting            |
| `df.unique()`          | Unique values           |
| `df.nunique()`         | Unique count            |
| `df.value_counts()`    | Values ki frequency     |
| `df.replace()`         | Value replace           |
| `df.astype()`          | Data type change        |
| `df.groupby()`         | Data grouping           |
| `df.mean()`            | Average                 |
| `df.sum()`             | Total                   |
| `df.max()`             | Maximum                 |
| `df.min()`             | Minimum                 |
| `df.count()`           | Count                   |
| `df.query()`           | Condition filtering     |


```python
import pandas as pd                                                    #CSV file ko Pandas me open/read karne ke liye use hota hai.

dis = {"a":[1,2,3,4,5,6],"s":[1,2,3,4,5,6],"d":[1,2,3,4,5,6]}

d = pd.DataFrame(dis)

print(d)
d.to_csv("Test_new.csv")
```

```python
import pandas as pd                                                   #df.to_csv--> #DataFrame ko CSV file me save karta hai.
                                                                       #index=False → extra index column save nahi karega.
dis = {"a":[1,2,3,4,5,6],"s":[1,2,3,4,5,9],"d":[1,2,3,4,5,6]}

d = pd.DataFrame(dis)

print(d)
d.to_csv("Test_new1.csv",index=False,header=[1,2,3])
```
