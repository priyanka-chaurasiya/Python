```python
# Join Array
```

```python
import numpy as np           # 1D array

var = np.array([1,2,3,4])
var1 = np.array([9,8,7,6])

ar = np.concatenate((var,var1))
print(ar)
```

```python
vr = np.array([[1,2],[3,4]])         # 2D array
vr1 = np.array([[9,8],[7,6]])

ar = np.concatenate((vr,vr1))
print(ar)
```

```python
vr = np.array([[1,2],[3,4]])
vr1 = np.array([[9,8],[7,6]])

print(vr)
```

```python
vr = np.array([[1,2],[3,4]])
vr1 = np.array([[9,8],[7,6]])

ar_new = np.concatenate((vr,vr1),axis=1)
print(vr)
print()
print(var1)
print()
print(ar_new)
```

```python
var = np.array([1,2,3,4])
var1 = np.array([9,8,7,6])

ar = np.stack((var,var1),axis=1)
print(ar)
```

```python
var_1 = np.array([1,2,3,4])
var_2 = np.array([9,8,7,6])

a_new = np.stack((var_1,var_2),axis = 1)
print(a_new)
```

```python
var_1 = np.array([1,2,3,4])              # Row Merge
var_2 = np.array([9,8,7,6])

a_new = np.hstack((var_1,var_2))
print(a_new)
```

```python
var_1 = np.array([1,2,3,4])              # Column Merge
var_2 = np.array([9,8,7,6])

a_new = np.vstack((var_1,var_2))
print(a_new)
```

```python
var_1 = np.array([1,2,3,4])              # Height Merge
var_2 = np.array([9,8,7,6])

a_new = np.dstack((var_1,var_2))
print(a_new)
```

```python
var_1 = np.array([1,2,3,4])
var_2 = np.array([9,8,7,6])

a_new = np.stack((var_1,var_2),axis = 0)
a_new1 = np.hstack((var_1,var_2))       # Row
a_new2 = np.vstack((var_1,var_2))       # Columns
a_new3 = np.dstack((var_1,var_2))       # Height


print(a_new)
print()
print(a_new1)
print()
print(a_new2)
print()
print(a_new3)
```

```python
var_1 = np.array([1,2,3,4])
var_2 = np.array([9,8,7,6])

a_new = np.stack((var_1,var_2),axis = 1)
a_new1 = np.hstack((var_1,var_2))       # Row
a_new2 = np.vstack((var_1,var_2))       # Columns
a_new3 = np.dstack((var_1,var_2))       # Height


print(a_new)
print()
print(a_new1)
print()
print(a_new2)
print()
print(a_new3)
```
