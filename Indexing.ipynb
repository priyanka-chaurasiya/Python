```python
import numpy as np

var = np.array([9,8,7,6])
#               0,1,2,3
#               -4,-3,-2,-1

print(var[1])
print(var[-4])
```

```python
var1 = np.array([[9,8,7],[4,5,6]])

print(var1)
print(var1.ndim)
print()

print(var1)
```

```python
var1 = np.array([[9,8,7],[4,5,6]])

print(var1)
print(var1.ndim)
print()

print(var1[0,2])
```

```python
var2 = np.array([[[1,2],[6,7]]])

print(var2)
print(var2.ndim)
print()

print(var2)
```

```python
var2 = np.array([[[1,2],[6,7]]])

print(var2)
print(var2.ndim)
print()

print(var2[0,1])
```

```python
var2 = np.array([[[1,2],[6,7]]])

print(var2)
print(var2.ndim)
print()

print(var2[0,1,1])
```

```python
arr = np.array([[1, 2, 3],             #2D array
                [4, 5, 6],
                [7, 8, 9]])

print(arr[0, 1])
print(arr[2, 2])
print(arr[1, 0])
```
