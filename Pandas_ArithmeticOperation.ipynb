```python
import pandas as pd                                                         # df.insert(position, "column_name", value)
                                                                               #insert() → DataFrame me kisi particular position par new column add karta hai.

var = pd.DataFrame({"A":[1,2,3,4,5],"B":[5,6,7,8,9],"c":[11,12,13,14,15]})

var
```

| Function   | Use               |
| ---------- | ----------------- |
| `insert()` | New column add    |
| `drop()`   | Row/column delete |
| `axis=0`   | Row               |
| `axis=1`   | Column            |



INSERT → Add Column

DROP   → Delete Row/Column

```python
var1 = var.pop("B")              # drop() → DataFrame se row ya column delete karta hai.

var1                             # axis=1 → Column delete

                                 # axis=0 → Row delete
```

```python
# Row Delete

import pandas as pd

data = {
    "Name": ["Priyanka", "Rahul", "Aman"],
    "Age": [20, 21, 22]
}

df = pd.DataFrame(data)
                                    # Index 1 wali row delete:
df = df.drop(1, axis=0)

print(df)
```

```python
# Multiple Columns Delete                           # Ek saath multiple columns delete kar sakte hain.

df = df.drop(["Name", "Age"], axis=1)
```

```python
# Multiple Rows Delete                                 # Ek saath multiple rows delete kar sakte hain.

df = df.drop([0, 2], axis=0)
```

```python
# CSV file ke saath

#Agar aapke paas data.csv hai:

import pandas as pd

df = pd.read_csv("data.csv")

# Insert
df.insert(1, "Country", "India")

# Delete column
df = df.drop("Country", axis=1)

# Save
df.to_csv("new_data.csv", index=False)
```
