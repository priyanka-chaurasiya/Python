```python
import matplotlib.pyplot as plt
import numpy as np

x=["python","c","c++","java"]
y=[85,70,60,82]
z=[20,30,40,50]

width=0.2
p=np.arange(len(x))
p1=[j+width for j in p]            #This moves the second set of bars slightly to the right.

plt.xlabel("language",fontsize=14)
plt.ylabel("No.",fontsize=14)
plt.title("result",fontsize=14)

width=0.2

plt.bar(p,y,width,color="r",label="Popularity")
plt.bar(p1,z,width,color="y",label="Popularity1")

plt.legend()
plt.show()


# x  → names/categories
# y  → first bar values
# z  → second bar values
# width → bar size
# p → first bar positions
# p1 → second bar positions
# bar() → creates bars
# xticks() → shows category names
# legend() → shows labels
# show() → displays graph
```

```python
import matplotlib.pyplot as plt
import numpy as np

x=["python","c","c++","java"]
y=[85,70,60,82]
z=[20,30,40,50]

width=0.2
p=np.arange(len(x))
p1=[j+width for j in p]

plt.xlabel("language",fontsize=14)
plt.ylabel("No.",fontsize=14)
plt.title("result",fontsize=14)

width=0.2

plt.bar(p,y,width,color="r",label="Popularity")
plt.bar(p1,z,width,color="y",label="Popularity1")


plt.xticks(p+width/2,x,rotation=20)             # xticks:use for name change karne ke liye
plt.legend()
plt.show()
```
