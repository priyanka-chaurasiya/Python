```python
import pandas as pd

var1 = pd.DataFrame({'A':[1,2,3,4],'B':[10,20,30,40]})
var2 = pd.DataFrame({'C':[10,20,30,40],'D':[100,200,300,400]})

var1.join(var2)
```

```python
import pandas as pd

var1 = pd.DataFrame({'A':[1,2,3,4],'B':[10,20,30,40]},index=["a","b","c","d"])
var2 = pd.DataFrame({'C':[10,20],'D':[100,200]},index=["a","b"])

var2.join(var1)
```

```python
import pandas as pd

var1 = pd.DataFrame({'A':[1,2,3,4],'B':[10,20,30,40]},index=["a","b","c","d"])
var2 = pd.DataFrame({'C':[10,20],'D':[100,200]},index=["a","b"])

# var2.join(var1,how="left")
# var2.join(var1,how="right")
# var2.join(var1,how="outer")
var2.join(var1,how="inner")
```

```python
var1 = pd.DataFrame({'A':[1,2,3,4],'B':[10,20,30,40]},index=["a","b","c","d"])
var2 = pd.DataFrame({'C':[10,20],'B':[100,200]},index=["a","b"])

var2.join(var1,how="outer",lsuffix="_12")
```

```python
import pandas as pd

var1 = pd.DataFrame({'A':[1,2,3,4],'B':[10,20,30,40]},index=["a","b","c","d"])
var2 = pd.DataFrame({'C':[10,20],'B':[100,200]},index=["a","b"])

# var1.append(var2) # The .append() method is deprecated and removed
pd.concat([var1, var2]) # Use pd.concat() instead
```

```python
var1 = pd.DataFrame({'A':[1,2,3,4],'B':[10,20,30,40]})
var2 = pd.DataFrame({'C':[10,20],'B':[100,200]})

# var1.append(var2) # The .append() method is deprecated and removed
pd.concat([var1, var2]) # Use pd.concat() instead
```

```python
import pandas as pd

var1 = pd.DataFrame({'A':[1,2,3,4],'B':[10,20,30,40]})
var2 = pd.DataFrame({'C':[10,20],'B':[100,200]})

# var1.append(var2,ignore_index=True) # The .append() method is deprecated and removed
pd.concat([var1, var2], ignore_index=True) # Use pd.concat() instead with ignore_index
```
