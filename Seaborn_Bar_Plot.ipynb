```python
# Bar Plot
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

```

```python
var = sns.load_dataset("penguins")
var
```

```python
sns.barplot(data=var, x="island", y="bill_length_mm")
plt.show()
```

```python
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex")
plt.show()
```

```python
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",order=order_1)
plt.show()
```

```python
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",
            order=order_1,hue_order=["Female","Male"],ci=1,n_boot=2)
plt.show()
```

```python
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",order=order_1,
            hue_order=["Female","Male"],orient="v")
plt.show()
```

```python
order_1 = ["Dream", "Torgersen", "Biscoe"]
sns.barplot(data=var, x="bill_depth_mm", y="bill_length_mm", orient="h")
plt.show()
```

```python
order_1 = ["Dream", "Torgersen", "Biscoe"]
sns.barplot(x="bill_depth_mm", y="bill_length_mm", data=var,orient="v")
plt.show()
```

```python
order_1 = ["Dream", "Torgersen", "Biscoe"]
sns.barplot(x="island",y="bill_length_mm",data=var,color="m")
plt.show()
```

```python
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",order=order_1,
            hue_order=["Female","Male"],orient="v",palette='icefire')
plt.show()
```

```python
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",order=order_1,
            hue_order=["Female","Male"],orient="v",saturation=10,errcolor="y")
plt.show()
```

```python
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",order=order_1,
            hue_order=["Female","Male"],orient="v",saturation=10,errcolor="y",errwidth=10)
plt.show()
```

```python
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",order=order_1,
            hue_order=["Female","Male"],orient="v",saturation=10,errcolor="b",capsize=0.5)
plt.show()
```

```python
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",order=order_1,
            hue_order=["Female","Male"],orient="v",saturation=100,errcolor="b",alpha=0.8)
plt.show()
```

```python
sns.set(style="darkgrid")
order_1 =["Dream","Torgersen","Biscoe"]
sns.barplot(data=var, x="island", y="bill_length_mm",hue="sex",order=order_1,
            hue_order=["Female","Male"],orient="v",saturation=100,errcolor="b",alpha=0.8)
plt.show()
```
