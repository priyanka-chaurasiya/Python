Missing Values


| Function    | Definition                  |
| ----------- | --------------------------- |
| `isnull()`  | Missing values check        |
| `isna()`    | Missing values check        |
| `notnull()` | Non-missing values check    |
| `dropna()`  | Missing rows/columns delete |
| `fillna()`  | Missing values fill         |


```python
import pandas as pd

var = pd.read_csv("/content/missing_data.csv")
var
```

```python
var.dropna()
```

```python
var.dropna(axis=1)
```

```python
import pandas as pd

var = pd.read_csv("/content/missing_data.csv")
var
```

```python
var.dropna(how="any")
```

```python
var.dropna(how="all")
```

```python
var.dropna(subset=["Email"])
```

```python
var.dropna(inplace=True)
var
```

```python
var.dropna(thresh=8)
var
```

```python
# fillna
```

```python
import pandas as pd

var = pd.read_csv("/content/missing_data.csv")
var
```

```python
var.fillna("python")
```

```python
import pandas as pd

var = pd.read_csv("/content/missing_data.csv")
var.fillna({"Employee_ID":"Python","Name":"c","Age":"java"})
```

```python
var.fillna(method="ffill")      # Previous value
```

```python
var.fillna(method="bfill")         # Next value
```

```python
var.fillna(method="bfill",axis=1)
```

```python
var.fillna(5,inplace=True)
var
```

```python
var.fillna("python",limit=2)
```
