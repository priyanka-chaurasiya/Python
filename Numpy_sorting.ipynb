```python
import numpy as np

x = np.array([[12,11,15],[21,25,20],[18,27,16]])

y = np.sort(x)
print(y)
```

```python
import numpy as np

x = np.array([[12,11,15],[21,25,20],[18,27,16]])

y = np.sort(x,axis=0)
print(y)
```

```python
import numpy as np

x = np.array([[12,11,15],[21,25,20],[18,27,16]])

y = np.argsort(x, axis=0)
print(y)
```

```python
import numpy as np

x = np.array([[12,11,15],[21,25,20],[18,27,16]])

y = np.argsort(x, axis=1)
print(y)
```

```python
import numpy as np

x = np.array([[12,11,15],[21,25,20],[18,27,16]])

x.sort()
print(x)
```

```python
import numpy as np         #1D array

x = np.array([2,3,4,5,6])

y = np.sort(x)
print(x)
```

```python
import numpy as np         #1D array

x = np.array([2,3,8,5,6])

y = np.sort(x)[::-1]
print(y)
```

```python
import numpy as np         #1D array

x = np.array([7,3,5,9,6])             # arrang the elemrnt in assending order
                                       # 3,5,6,7,9
                                                # 7,3,5,9,6
y = np.argsort(x)                       #indexing 0,1,2,3,4
print(y)
```
