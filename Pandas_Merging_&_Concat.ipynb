```python
# Merge
```

```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
var2 = pd.DataFrame({"A":[1,2,3,5],
                    "C":[8,6,3,9]})
pd.merge(var1,var2,on="A")
```

```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
var2 = pd.DataFrame({"A":[1,2,3,5],
                    "C":[8,6,3,9]})
pd.merge(var1,var2,how = "left")
```

```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
var2 = pd.DataFrame({"A":[1,2,3,5],
                    "C":[8,6,3,9]})
pd.merge(var1,var2,how = "right")
```

```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
var2 = pd.DataFrame({"A":[1,2,3,5],
                    "C":[8,6,3,9]})
pd.merge(var1,var2,how = "outer")
```

```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
var2 = pd.DataFrame({"A":[1,2,3,5],
                    "C":[8,6,3,9]})
pd.merge(var1,var2,how = "outer",indicator=True)
```

```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
var2 = pd.DataFrame({"A":[1,2,3,5],
                    "B":[8,6,3,9]})
pd.merge(var1,var2,left_index=True,right_index=True)
```

```python
import pandas as pd

var1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
var2 = pd.DataFrame({"A":[1,2,3,5],
                    "B":[8,6,3,9]})
pd.merge(var1,var2,left_index=True,right_index=True,suffixes=("name","python"))
```

```python
# Concat
```

```python
sr1 = pd.Series([1,2,3,4])
sr2 = pd.Series([11,12,31,41])
pd.concat([sr1,sr2])
```

```python
sr1 = pd.Series([1,2,3,4])
sr2 = pd.Series([11,12,31,41])

pd.concat([sr1,sr2])
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
d2 = pd.DataFrame({"A":[1,2,3,5],
                    "B":[8,6,3,9]})

pd.concat([d1,d2])
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
d2 = pd.DataFrame({"A":[1,2,3,5],
                    "C":[8,6,3,9]})
pd.concat([d1,d2],axis=1)
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
d2 = pd.DataFrame({"A":[1,2,],
                    "C":[8,6]})
pd.concat([d1,d2],axis=1,join="inner")
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
d2 = pd.DataFrame({"A":[1,2,3,4],
                    "C":[21,32,31,11]})
pd.concat([d1,d2],axis=0,join="outer",keys=["first","second"])
```

```python
d1 = pd.DataFrame({"A":[1,2,3,4],
                    "B":[5,6,7,8]})
d2 = pd.DataFrame({"A":[1,2,3,4],
                    "C":[21,32,31,11]})
pd.concat([d1,d2],axis=1,keys=["d1","d2"])
```

```python
d1 = pd.DataFrame({"D":[1,2,3,4]})
d2 = pd.DataFrame({"A":[1,2,3,4],"C":[21,32,31,11]})

pd.concat([d1,d2])
```
