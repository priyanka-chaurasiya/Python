```python
import matplotlib.pyplot as plt


x=[1,2,3,4,5]
y=[3,1,5,2,6]

plt.plot(x,y)

plt.title("My First Graph",fontsize=15)
plt.xlabel("Days",fontsize=15)
plt.ylabel("Python",fontsize=15)

plt.show()
```

```python
import matplotlib.pyplot as plt


x=[1,2,3,4,5]
y=[3,1,5,2,6]

plt.plot(x,y)

plt.title("My First Graph",fontsize=15)
plt.xlabel("Days",fontsize=15)
plt.ylabel("Python",fontsize=15)

plt.text(2,3,"Java",fontsize=10,style="italic",color="red")
plt.text(3,1,"C++",fontsize=10,style="italic",color="green")
plt.text(4,5,"C",fontsize=10,style="italic",color="blue")
plt.show()
```

```python
import matplotlib.pyplot as plt


x=[1,2,3,4,5]
y=[3,1,5,2,6]

plt.plot(x,y)

plt.title("My First Graph",fontsize=15)
plt.xlabel("Days",fontsize=15)
plt.ylabel("Python",fontsize=15)

plt.text(2,3,"Java",fontsize=10,style="italic",bbox={"facecolor":"m"})
plt.annotate("python",xy=(2,1),xytext=(3,3),arrowprops=dict(facecolor="black",shrink=100))
plt.show()
```

```python
import matplotlib.pyplot as plt


x=[1,2,3,4,5]
y=[3,1,5,2,6]

plt.plot(x,y)

plt.title("My First Graph",fontsize=15)
plt.xlabel("Days",fontsize=15)
plt.ylabel("Python",fontsize=15)

plt.text(2,3,"Java",fontsize=10,style="italic",bbox={"facecolor":"m"})

plt.annotate("python",xy=(2,1),xytext=(3,3),arrowprops=dict(facecolor="black",shrink=100))

plt.legend(["Python"],loc=9,facecolor="r",edgecolor="yellow",framealpha=0.5,shadow=True)

plt.show()
```
