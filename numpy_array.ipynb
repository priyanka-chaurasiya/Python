```python
# Arithmetic operation
```

```python
import numpy as np

var = np.array([1,2,3,4])

varadd = var + 3

print(varadd)
```

```python
var1 = np.array([1,2,3,4])         # add to var.
var2 = np.array([1,2,3,4])

varadd = var1 + var2

print(varadd)
```

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])

varadd = np.add(var1,3)

print(varadd)
```

```python
var = np.array([1,2,3,4])

varadd = var - 3

print(varadd)
```

```python
var = np.array([1,2,3,4])

varadd = var * 3

print(varadd)
```

```python
var = np.array([1,2,3,4])

varadd = var % 3

print(varadd)
```

```python
var = np.array([1,2,3,4])

varadd = var / 3

print(varadd)
```

```python
var = np.array([1,2,3,4])

varadd = var ** 3

print(varadd)
```

```python
var = np.array([1,2,3,4])

varadd = np.reciprocal(var)

print(varadd)
```
