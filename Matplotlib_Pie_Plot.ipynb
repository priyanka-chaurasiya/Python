```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=["c","c++","java","python","javascript"]
ex=[0.4,0.0,0.0,0.0,0.0]

plt.pie(x,labels=y,explode=ex)
plt.show()
```
