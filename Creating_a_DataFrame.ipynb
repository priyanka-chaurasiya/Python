```python
import pandas as pd

std_data=[(1, 'varun', 30, 'male', 'chandigarh'),
          (2, 'ravi', 31, 'male', 'delhi'),
          (3, 'preeti', 29, 'female', 'jaipur'),
          (4, 'amrit', 32, 'male','mumbai'),
          (5, 'priyanka', 20, 'female','banglore')]

df=pd.DataFrame(std_data, columns=['id', 'name', 'age', 'gender', 'city'])
df
```

```python
import pandas as pd

l=[1,2,3,4,5,6]

var = pd.DataFrame(l)

print(type(var))
```

```python
d = {"a":[1,2,3,4,5],"s":[1,2,3,4,5]}

var1 = pd.DataFrame(d)

print(type(var1))
print(var1)
```

```python
d = {"a":[1,2,3,4,5],"s":[1,2,3,4,5]}

var1 = pd.DataFrame(d,columns=["a"])

print(type(var1))
print(var1)
```

```python
d = {"a":[1,2,3,4,5],"s":[1,2,3,4,5],"d":[1,2,3,4,5],1:[1,2,3,4,5]}

var1 = pd.DataFrame(d,columns=["a",1],index=["a","s","d","f","g"])

print(type(var1))
print(var1)
```

```python
d = {"a":[1,2,3,24,5],"s":[1,2,3,4,5],"d":[1,2,3,4,5],1:[1,2,3,4,5]}

var1 = pd.DataFrame(d)

print(type(var1))
print(var1)

print(var1["a"][3])
```

```python
list_1 = [[1,2,3,4,5],[11,12,13,14,15]]

var2 = pd.DataFrame(list_1)

print(type(var2))
print(var2)
```

```python
sr = {"s":pd.Series([1,2,3,4,5]),"r":pd.Series([1,2,3,4,5])}

var3 = pd.DataFrame(sr)

print(type(var3))
print(var3)
```
