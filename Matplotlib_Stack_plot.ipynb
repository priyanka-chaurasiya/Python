```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area = [1,2,3,4,5]

plt.stackplot(x,area)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area = [2,3,2,5,4]
area1 = [2,3,1,5,4]
area2 = [2,1,4,5,4]

plt.stackplot(x,area,area1,area2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area = [2,3,2,5,4]
area1 = [2,3,1,5,4]
area2 = [2,1,4,5,4]

l=["area","area1","area2"]
plt.stackplot(x,area,area1,area2,labels=l)
plt.legend(loc=2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area = [2,3,2,5,4]
area1 = [2,3,1,5,4]
area2 = [2,1,4,5,4]

l=["area","area1","area2"]
plt.stackplot(x,area,area1,area2,labels=l,colors=["y","g","m"])
plt.legend(loc=2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area = [2,3,2,5,4]
area1 = [2,3,1,5,4]
area2 = [2,1,4,5,4]

l=["area","area1","area2"]

plt.stackplot(x,area,area1,area2,labels=l,colors=["y","g","m"],baseline="sym") #wiggle
plt.legend(loc=2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area = [2,3,2,5,4]
area1 = [2,3,1,5,4]
area2 = [2,1,4,5,4]

l=["area","area1","area2"]
plt.stackplot(x,area,area1,area2,labels=l,colors=["y","g","m"])

plt.title("python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend(loc=2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area = [2,3,2,5,4]
area1 = [2,3,1,5,4]
area2 = [2,1,4,5,4]

l=["area","area1","area2"]
plt.stackplot(x,area,area1,area2,labels=l,colors=["y","g","m"])

plt.title("python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.legend(loc=2)
plt.show()
```
