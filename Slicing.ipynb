```python
import numpy as np

var = np.array([1,2,3,4,5,6,7])
#               0,1,2,3,4,5,6

print(var)

print()

print("2 to 5 :",var[1:5])
```

```python
import numpy as np

var = np.array([1,2,3,4,5,6,7])
#               0,1,2,3,4,5,6

print(var)

print()

print("2 to 5 :",var[1:5])

print("2 to End :",var[1:])
```

```python
import numpy as np                  #2D array

array1 = np.array([[15,16,17],[25,26,27],[35,36,37],[45,46,47]])

print(array1[1,])
print(array1[:,1])
print(array1[1:3,1:3])
print(array1[1:3, ])
print(array1[: ,1:3])
print(array1[1:3,1])
print(array1[1:3,:1])
```
