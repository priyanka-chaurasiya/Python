```python
# Box Plot
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]

plt.boxplot(x)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]

plt.boxplot(x,notch=True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]

plt.boxplot(x,vert=False)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]

plt.boxplot(x,widths=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]

plt.boxplot(x,labels=["Python"])
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]

plt.boxplot(x,labels=["Python"],patch_artist=True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]

plt.boxplot(x,labels=["Python"],showmeans=True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70,120,200]

plt.boxplot(x,labels=["Python"],showmeans=True,
            whis=2)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70,120,200]

plt.boxplot(x,labels=["Python"],showmeans=True,
            sym="g+",boxprops=dict(color="red"),capprops=dict(color="green"))
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70,120,200]

plt.boxplot(x,labels=["Python"],showmeans=True,
            sym="g+",boxprops=dict(color="red"),capprops=dict(color="green"),
            whiskerprops=dict(color="b"),flierprops=dict(color="y"))
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]
x1=[10,20,40,50,30,60,90]

plt.boxplot(x,labels=["Python"],showmeans=True,
            sym="g+",boxprops=dict(color="red"),capprops=dict(color="green"),
            whiskerprops=dict(color="b"))

plt.boxplot(x1,labels=["Python"],showmeans=True,
            sym="g+",boxprops=dict(color="red"),capprops=dict(color="green"),
            whiskerprops=dict(color="b"))
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]
x1=[10,20,40,50,30,60,90]

y=[x,x1]
plt.boxplot(y,showmeans=True,
            sym="g+",boxprops=dict(color="red"),capprops=dict(color="green"),
            whiskerprops=dict(color="b"))

plt.show()
```

```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70]
x1=[10,20,40,50,30,60,90]

y=[x,x1]
plt.boxplot(y,labels=["python","c++"],showmeans=True,
            sym="g+",boxprops=dict(color="red"),capprops=dict(color="green"),
            whiskerprops=dict(color="b"))

plt.show()
```
