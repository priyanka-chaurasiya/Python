```python
# Melt()
```

data ko reshape karne ke liye use karte hai melt() and pivot function

```python
import pandas as pd

var = pd.DataFrame({"days":[1,2,3,4,5,6],"eng":[10,12,14,15,16,12],"math":[17,18,19,13,14,16]})

var
```

```python
pd.melt(var)
```

```python
pd.melt(var,id_vars=["days"])
```

```python
pd.melt(var,id_vars=["days"],var_name="python")
```

```python
pd.melt(var,id_vars=["days"],var_name="python",value_name="wsjob")
```

```python
# Pivot Function #Pivot()
```

```python
import pandas as pd

var = pd.DataFrame({"days":[1,2,3,4,5,6],
                    "st_name":["a","b","c","a","b","c"],
                    "eng":[10,12,14,15,16,12],"math":[17,18,19,13,14,16]})

var
```

```python
var.pivot(index="days",columns="st_name")
```

```python
import pandas as pd

var = pd.DataFrame({"days":[1,2,3,4,5,6],
                    "st_name":["a","b","b","a","b","b"],
                    "eng":[10,12,14,15,16,12],"math":[17,18,19,13,14,16]})

var
```

```python
var.pivot(index="days",columns="st_name")
```

```python
var.pivot(index="days",columns="st_name",values="eng")
```

```python
import pandas as pd

var = pd.DataFrame({"days":[1,1,1,1,2,2],
                    "st_name":["a","b","b","a","b","b"],
                    "eng":[10,12,14,15,16,12],"math":[17,18,19,13,14,16]})

var
```

```python
var.pivot_table(index="days",columns="st_name",aggfunc="sum")
```

```python
var.pivot_table(index="days",columns="st_name",aggfunc="mean")
```

```python
var.pivot_table(index="days",columns="st_name",aggfunc="sum",margins=True)
```
