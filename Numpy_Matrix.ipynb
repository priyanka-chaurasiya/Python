```python
import numpy as np

var = np.matrix([[1,2,3],[1,2,3]])

print(var)
print(type(var))
```

```python
var1 = np.array([[1,2,3],[1,2,3]])

print(var1)
print(type(var1))
```

```python
import numpy as np

var = np.matrix([[1,2,3],[1,2,3]])
var2 = np.matrix([[1,2,3],[1,2,3]])

print(var)
print(type(var))
print()
print(var+var2)
```

```python
import numpy as np

var = np.matrix([[1,2,3],[1,2,3]])
var2 = np.matrix([[1,2,3],[1,2,3]])

print(var)
print(type(var))
print()
print(var-var2)
```

```python
import numpy as np

var = np.matrix([[1,2],[1,2]])
var2 = np.matrix([[1,2],[1,2]])

print(var)
print(type(var))
print()
print(var*var2)
```

```python
import numpy as np

var = np.matrix([[1,2],[1,2]])
var2 = np.matrix([[1,2],[1,2]])

print(var)
print(type(var))
print()
print(var*var2)
print(var.dot(var2))
```

```python
var1 = np.array([[1,2,3],[1,2,3]])

print(var1)
print(type(var1))
print()
print(var1*var1)
```

```python
# Matrix Function
```

```python
import numpy as np

var = np.matrix([[1,2,3],[4,5,6]])
print(var)
print()
print(np.transpose(var))
```

```python
import numpy as np

var = np.matrix([[1,2,3],[4,5,6]])
print(var)
print()
print(np.transpose(var))

print()
print(var.T)
```

```python
import numpy as np

var = np.matrix([[1,2,3],[4,5,6]])
print(var)
print()
print(np.transpose(var))

print()
print(var.T)
print()
print(np.swapaxes(var,0,1))
```

```python
var2 = np.matrix([[1,2],[3,4]])
print(var2)
print()
print(np.swapaxes(var2,0,1))
```

```python
var3 = np.matrix([[1,2],[3,4]])
print(var3)
print()
print(np.linalg.inv(var3))
```

```python
var4 = np.matrix([[1,2,3],[4,5,6],[7,8,9]])
print(var4)
print()
print(np.linalg.matrix_power(var4,2))
```

```python
var5 = np.matrix([[1,2,3],[1,5,5],[3,8,4]])
print(var5)
print()
print(np.linalg.det(var5))
```
