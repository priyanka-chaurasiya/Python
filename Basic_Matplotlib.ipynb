```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[2,4,6,8]

plt.plot(x,y)

plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[2,4,6,8]

plt.bar(x,y)

plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[2,4,6,8]

c=["r","y","g","b"]

plt.bar(x,y,color=c)

plt.show()
```
