```python
#Step plot
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[1,2,3,4,5]

plt.step(x,y)

plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[1,5,3,4,6]

plt.step(x,y,color="b",marker="o",ms=10,mfc="m")

plt.title("Step Plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[1,5,3,4,6]

plt.step(x,y,color="b",marker="o",ms=10,mfc="m",label="python")

plt.title("Step Plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.legend()
plt.grid()
plt.show()
```
